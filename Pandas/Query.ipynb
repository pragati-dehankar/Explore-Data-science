```python
#rules and tips

#1. Column names become variables
df.query("Age > 22 and City == 'Delhi'")
```

```python
#2. String values must be in quotes

df.query("City == 'Delhi'")
```

```python
#3. Use backticks for column names with spaces or special characters

df.query("`Score` == 58")
```

```python
#4. You can use @ to reference Python variables
age_limit = 22
df.query("Age > @age_limit")
```

```python
#5. Logical operators (and,or,not)
df.query("Age ==19 and City == 'Delhi'") 
```

```python
#6. Chained comparisons
df.query("19 < Age <= 22")
```

```python
#7. .query() returns a copy, not a view

filtered = df.query("Age < 21")
filtered
```
