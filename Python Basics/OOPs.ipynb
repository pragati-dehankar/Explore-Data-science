```python
class student:
    def __init__(self,Name,Age):
        self.Name=Name
        self.Age=Age
    def display(self):
        return f"{self.Name} {self.Age}"

s=student("Pragati",21)
s.display()
```

```python
#Encapsulation
class std:
    def __init__(self,nam,kam):
        self.name=nam
        self.__kam=kam        #private
    def get_kam(self):
        return self.__kam

    def set_kam(self,kam):
        self.__kam=kam

s=std("pragati",2)
print(s.get_kam())
s.set_kam("kuch nh ")
print(s.get_kam())
```

```python
#Inheritance
class Person:
    def __init__(self,name):
        self.name=name
    def show_name(self):
        print("Name:",self.name)
class Teacher(Person):
    def __init__(self,name,subject):
        super().__init__(name)
        self.subject=subject
    def show_subject(self):
        print("subject:",self.subject)

t=Teacher("Pragati","Maths")
t.show_name()
t.show_subject()
```

```python
#Polymorphism

class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")

d=Dog()
c=Cat()

d.sound()
c.sound()
```

```python
#Dunder methods

class Book:
    def __init__(self,title,pages):
        self.title=title
        self.pages=pages
    def __str__(self):
        return f"Book:{self.title}"

    def __len__(self):
        return self.pages

b=Book("Python",345)
print(str(b))
print(len(b))
```
