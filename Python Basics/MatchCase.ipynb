```python
# Input a character → check vowel or consonant
# print (ch[0])
ch=input("Enter a character").lower()
match ch:
    case 'a'|'e'|'i'|'o'|'u':
            print("Vowel")
    case _:
        if len(ch)==1 and ch.isalpha():
                print ("consonent")
        else:
                print ("Enter proper character")
           
```

```python
num=int(input("enter a num"))
match num:
    case 1:
        print("sunday")
    case 2:
        print("Monady")
    case 3:
        print("Tuesday")
    case 4:
        print("Wednesday")
    case 5:
        print("thursday")
    case 6:
        print("Friday")
    case 7:
        print("Saturday")
    
```
