```python
#Reverse a number
num=int(input("enetr a num"))
rev=0
while num!=0:
    dig=num%10
    rev=rev*10+dig
    num//=10
print(rev)
```

```python
#Number guessing game
import random

secret=random.randint(1,100)
attempt=0

while True:
    guess=int(input("enter a num bet 1-100: "))
    attempt+=1

    if guess<secret: 
        print("small num guess kiye tumne ")
    elif guess>secret:
        print("bada number guess kiye tumne")
    else:
        print("Correct number you Guessed in",attempt,"attempts")
        break
    
```
