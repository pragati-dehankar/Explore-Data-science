```python
#Serialization

import json

std={"id": 101,
    "name": "Pragati",
    "age": 22,
    "course": "Computer Science",
    "skills": ["Python", "React", "AI"],
    "email": "pragati@example.com",
    "isStudent": True,
    "city": "Nagpur"}
jsonStr=json.dumps(std)
print(jsonStr)
print(type(jsonStr))
```

```python
with open("data.json","w") as file:
    json.dump(std,file)
```

```python
#Deserialization
import json

jsonData='{"name":"praachi","age":21,"city":"ballarpur"}'
pyObj=json.loads(jsonData)
print(pyObj)
print(type(pyObj))
```

```python
with open("data.json","r") as file:
    pyData=json.load(file)
print(pyData)
```
