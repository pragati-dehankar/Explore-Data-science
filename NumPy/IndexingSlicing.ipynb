```python
import numpy as np

arr=np.array([12,13,14,56,78,90])
print(arr)
print(arr[0])
print(arr[4])
```

```python
print(arr[1:2])
print(arr[:5])
print(arr[2:])
```

```python
print(arr[::4])
```

```python
s=arr[1:3]
s[0]=23445
print(s)
print(arr)
```

```python
#Fancy Indexing 

ar=np.array([10,20,30,40])
idx=[2,3,0]
print(ar[idx])
```

```python
#Boolean Masking 
a=np.array([34,45,56,67,21,12,43,90,10])
mask=a>=45
print(mask)
a[mask]
```
