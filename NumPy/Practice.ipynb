```python
#Compare the time taken to multiply two Python lists vs. two NumPy arrays
import numpy as np
import time

size = 1000000
list1 = list(range(size))
list2 = list(range(size))

arr1 = np.array(list1)
arr2 = np.array(list2)

start=time.time()
res_list=[a*b for a,b in zip(list1,list2)]
end=time.time()
print("List time:",end-start)

start=time.time()
res_arr=arr1*arr2
end=time.time()
print("NumPy time:", end - start)
```

```python
# find the memory size of a NumPy array with 1 million elements.

import numpy as np

arr=np.arange(1000000)

print("Memory size (bytes):", arr.nbytes)
```

```python
#Convert an array of floats [1.1, 2.2, 3.3] into integers.
floatArr=np.array([1.1, 2.2, 3.3])
print(floatArr.astype(np.int32))
```

```python
#Use fancy indexing to extract even numbers from [1, 2, 3, 4, 5, 6]
d=np.array([1, 2, 3, 4, 5, 6])
i=d[d%2==0]
print(i)
```
