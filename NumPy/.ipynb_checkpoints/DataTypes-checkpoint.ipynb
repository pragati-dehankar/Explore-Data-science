```python
import numpy as np

arr=np.array([12.3,34.6])
print(arr.dtype)
```

```python
arrInt=arr.astype(np.int32)
print(arrInt)
print(arrInt.dtype)
```

```python
#Downcasting
arr_large = np.array([1000000, 2000000, 3000000], dtype=np.int64)
arr_small = arr_large.astype(np.int32)
print(arr_small.dtype)
print(arr_large.dtype)
```

```python
#Memory usage
arr_int64 = np.array([1, 2, 3], dtype=np.int64)
arr_int32 = np.array([1, 2, 3], dtype=np.int32)
print(arr_int32.nbytes)
print(arr_int64.nbytes)
```

```python
# String Data Type in NumPy
a=np.array(['apple','banana','strawberry'])
print(a.dtype)
```

```python
#Complex
arr = np.array([1 + 2j, 3 + 4j, 5 + 6j])
print(arr.dtype)
```

```python
#Object Data Type
ar = np.array([{'b': 56}, [1, 2, 3], 'hii'],dtype=object)
print(ar)
```
